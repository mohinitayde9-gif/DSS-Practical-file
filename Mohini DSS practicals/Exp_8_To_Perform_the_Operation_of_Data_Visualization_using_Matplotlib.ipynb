```python
#importing the basic library
import numpy as np
import matplotlib as plt
from matplotlib import pyplot as plt # pyplot for python,ggplot for r language
```

```python
x=np.arange(1,11)
x
```

```python
y=2*x
y
```

```python
# line chart
plt.plot(x,y)
plt.title('line chart')
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

```python
# bar chart
plt.bar(x,y)
```

```python
# bar chart
plt.bar(x,y)
plt.title('bar chart')
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

```python
# scatter plot
a = (5,4,2,9,1,8,2,1,8,5)
b = (10,52,45,33,36,25,45,24,14,28)
plt.scatter(a,b)
plt.title('scatter chart')
plt.xlabel("a axis")
plt.ylabel("b axis")
plt.show()
```

```python
plt.scatter(x,y)
plt.title('scatter chart')
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```
